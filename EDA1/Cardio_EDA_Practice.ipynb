```python
import pandas as pd
import seaborn as sns
```

```python
df.info()
```

```python
#we do have missing values in here in this data 
df.isnull().sum()
```

```python
#iam dropping this missing values because 
#This missing values are less than 1 percent of the data we have
df.dropna(inplace=True)
```
