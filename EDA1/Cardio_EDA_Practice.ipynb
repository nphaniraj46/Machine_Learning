```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("Cardiotocographic.csv")
```

```python
df=pd.DataFrame(data)
df.sample(5) #This gives me some random rows data
#if we use this df.head() we wi;; get only first 5 rows data
```

```python
df.info()
```

```python
#we do have missing values in here in this data 
df.isnull().sum()
```

```python
#iam dropping this missing values because 
#This missing values are less than 1 percent of the data we have
df.dropna(inplace=True)
```

```python
df.shape #we have about 2123 rows of data
```

```python
df.isnull().sum() #null are removed
```

```python
df.describe()
#This gives me statistical summary for every numerical row
```
